```python
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt

import detector_model as dm
```

```python
detector = dm.Detector()

print(detector.pmt_positions.shape)

assert np.allclose(
    np.linalg.norm(detector.pmt_positions, axis=1),
    detector.R
)

assert np.array_equal(
    detector.find_pmt(detector.pmt_positions),
    np.arange(detector.N)
)

print(
    "Попадание в северный полюс:",
    detector.find_pmt([0, 0, detector.R])
)
```

```python
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(projection="3d")

positions = detector.pmt_positions

ax.scatter(
    positions[:, 0],
    positions[:, 1],
    positions[:, 2],
    s=50
)

ax.set_xlabel("x, м")
ax.set_ylabel("y, м")
ax.set_zlabel("z, м")

ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.set_zlim(-2, 2)

ax.set_box_aspect((1, 1, 1))

plt.show()
```
